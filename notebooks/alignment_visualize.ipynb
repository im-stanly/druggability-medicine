```python
import Bio.PDB
import py3Dmol
import io
```

```python
parser = Bio.PDB.MMCIFParser(QUIET=True)

str_exp = parser.get_structure("exp", "../data/03_primary/matched_protein_pairs/A0A8C6VAB1/PDB.cif")
str_af = parser.get_structure("af", "../data/03_primary/matched_protein_pairs/A0A8C6VAB1/AF.cif")
```

```python
exp_model = str_exp[0]
af_model = str_af[0]


exp_atoms = [atom for atom in exp_model.get_atoms() if atom.get_name() == 'CA']
af_atoms = [atom for atom in af_model.get_atoms() if atom.get_name() == 'CA']
```

```python
superimposer = Bio.PDB.Superimposer()
superimposer.set_atoms(exp_atoms, af_atoms)
superimposer.apply(af_model.get_atoms())
print(f"Alignment complete! RMSD: {superimposer.rms:.3f} Å")
```

```python
io_obj = Bio.PDB.PDBIO()

io_obj.set_structure(str_exp)
f1 = io.StringIO()
io_obj.save(f1)
pdb_str_exp = f1.getvalue()


io_obj.set_structure(str_af)
f2 = io.StringIO()
io_obj.save(f2)
pdb_str_af = f2.getvalue()
```

```python
view = py3Dmol.view(width=800, height=500)

dir =  "../data/04_feature/aligned_protein_pairs/"
name = "P01112"

with open(f"{dir}{name}/PDB.cif") as f:
    view.addModel(f.read(), 'cif')
    view.setStyle({'model': 0}, {'cartoon': {'colorscheme': 'cyanCarbon'}})

with open(f"{dir}{name}/AF.cif") as f:
    view.addModel(f.read(), 'cif')
    view.setStyle({'model': 1}, {'cartoon': {'colorscheme': 'magentaCarbon'}})

view.zoomTo()
view.show()
```
